## If something goes wrong

- **No Connect / can't run cells:** Sign in, refresh the page, and click **Connect**. Use a standard CPU runtime.
- **Install error:** Rerun Step 1. If Colab asks for a runtime restart, do that and run all cells again.
- **`ModuleNotFoundError`:** Rerun Step 1, then Step 2.
- **Download / network error in Step 3:** Check your internet connection and retry. Some institutional networks restrict downloads.
- **`NameError` (such as `adata` not defined):** Choose **Runtime → Run all** to execute cells in order.
- **Any other error:** Share the **step number** and the **error text or screenshot** with the instructor.
